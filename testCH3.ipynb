## Page 7

Let's use the Hugging Face transformers library.

For this example, we will load distilgpt2, a small pretrained model for generating English text.

```python
# pip install torch transformers
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

# Select a pretrained model
model_name = "distilgpt2"

# Load the tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Load the model architecture and trained weights
model = AutoModelForCausalLM.from_pretrained(model_name)

print("Model loaded successfully!")
  
```

```python

prompt = "Data science is"

# Convert text into token IDs
inputs = tokenizer(prompt, return_tensors="pt")

# Generate more tokens
outputs = model.generate(
    **inputs,
    max_new_tokens=30,
    do_sample=False,
    pad_token_id=tokenizer.eos_token_id
)

# Convert generated token IDs back into text
result = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(result)
```

```python

import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# 1. Select the model
model_name = "gpt2"

# 2. Load the tokenizer and model
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

# 3. Prepare the input
prompt = "Data science is"
inputs = tokenizer(prompt, return_tensors="pt")

# 4. Generate text
model.eval()

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

# 5. Convert token IDs back to text
result = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(result)
```

```python

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# 1. Choose a pretrained model
model_name = "gpt2"

# 2. Load the tokenizer automatically
tokenizer = AutoTokenizer.from_pretrained(model_name)

# 3. Load the model architecture automatically
model = AutoModelForCausalLM.from_pretrained(model_name)

# 4. Prepare the text
prompt = "Machine learning is"
inputs = tokenizer(prompt, return_tensors="pt")

# 5. Generate text
model.eval()

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=30,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

# 6. Convert token IDs into text
result = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(result)
```
